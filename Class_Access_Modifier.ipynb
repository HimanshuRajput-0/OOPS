```python
class Human:
          def __init__ (self, name, age):
                self.HumanName = name
                self.HumanAge = age
          def display(self):
                print("name:", self.HumanName, "and", "age :" , self.HumanAge)

object = Human("Himanshu",24)
object.display()
```
